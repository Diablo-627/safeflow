```python
import pandas as pd
import numpy as np
import os

# Симулируем "сырые" данные от сенсоров
data = []
for device_id in range(1, 6):  # 5 устройств
    for t in range(50):  # 50 измерений на устройство
        data.append({
            "timestamp": t,
            "device_id": device_id,
            "sensor_1": np.random.randint(5, 25),
            "sensor_2": np.random.randint(2, 15),
            "sensor_3": np.random.randint(1, 10),
        })

raw_df = pd.DataFrame(data)
raw_df.head()

```

```python
# Создаём фичи — средние значения по каждому устройству
features = (
    raw_df
    .groupby("device_id")
    .agg({
        "sensor_1": "mean",
        "sensor_2": "mean",
        "sensor_3": "mean"
    })
    .reset_index()
)

# Добавляем целевую переменную (target)
features["target"] = (features["sensor_1"] > 15).astype(int)

# Проверяем распределение классов
print(features["target"].value_counts())

features.head()
```

```python
# Сохраняем итоговый датасет
os.makedirs("data", exist_ok=True)
features.to_csv("data/train_dataset.csv", index=False)
print("✅ train_dataset.csv сохранён в ml/data/")
```
